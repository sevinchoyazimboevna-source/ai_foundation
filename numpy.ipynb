```python
import numpy as np
arr = np.array([1,3,5,7])
print('numpy_array:', arr)
```

```python
arr2 = np.array([[1,3,4,6],[4,6,8,9]])
print('numpy_array:', arr2)
```

```python
#d type
a = np.array([1,2,3])
print(a.dtype)
```

```python
# reshape
a = np.array([1,2,3,4,5,6])
b = a.reshape(2,3)
print("shaped_array:", b)
```

```python
# astype()
a = np.array([1,2,3])
b = a.astype(float)
print("float:", b)
```

```python
# np.zeros(), np.ones() => dunny data
z = np.zeros((2,3))
o = np.ones((2,3))

print("zeroz", z)
print("ones", o)
```

```python
# ndin
a = np.array([[1,2], [3,4]])
print("ndim", a)
```

```python
# arange()
a = np.arange(1,7)
print("arange", a)
```

```python
# sum(),mean(), max(),min()
a = np.array([1,2,3,4])

print("sum",a.sum())
print("mean",a.mean())
print("max", a.max())
print("min",a.min())
```

```python
# flatten ()

a = np.array([[1,2], [3,4]])
print("flatten", a.flatten())
```

```python
# T(Transpose)
a = np.array([[1,2], [3,4]])
print("transpose", a.T)
```

```python
# dot()
a = np.array([[1,2],[3,4]])
b = np.array([[2,0],[1,2]])

print("dot_array", a.dot(b))
```

```python
# random.rand()
a = np.random.rand(2,3) # =>0.0 and 1.0 / 0.0 included / 1.0 excluded
print("randomrand", a)
```
